```python

from itertools import permutations

def travelling_salesman(graph, start):
    n = len(graph)
    vertices = [i for i in range(n) if i != start]

    min_cost = float('inf')
    best_path = []

    for path in permutations(vertices):
        cost = 0
        current = start

        for vertex in path:
            cost += graph[current][vertex]
            current = vertex

        cost += graph[current][start]

        if cost < min_cost:
            min_cost = cost
            best_path = [start] + list(path) + [start]

    print("Minimum Cost:", min_cost)
    print("Best Path:", " -> ".join(map(str, best_path)))


# Example input
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

travelling_salesman(graph, 0)
```
